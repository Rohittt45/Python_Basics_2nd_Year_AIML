```python
sales_data = ["1200", "1500", "1100", "1800", "1600", "1400", "1700"]
sales_int=[]

for i in sales_data:
    if(type(i)=="int"):
        sales_int.append(i)
    if(type(i)!="int"):
        i=int(i)
        sales_int.append(i)

for i,x in enumerate(sales_int,start=1):
    print(f"Day{i}Sales Rs",x)

print(f"Total Sales Rs",sum(sales_int))
```
